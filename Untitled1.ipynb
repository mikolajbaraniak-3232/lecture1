```python
import pandas as pd
import matplotlib.pyplot as plt

# Wczytaj dane z pliku CSV
df = pd.read_csv('/content/Salary Data (2).csv')

# Wyświetl pierwsze 5 wierszy danych, aby zobaczyć ich strukturę
print("Pierwsze 5 wierszy danych:")
display(df.head())

# Sprawdź dostępność kolumn przed ich użyciem
if 'Years of Experience' in df.columns and 'Salary' in df.columns:
    # Stwórz wykres punktowy
    plt.figure(figsize=(10, 6))
    plt.scatter(df['Years of Experience'], df['Salary'])
    plt.title('Zależność między latami doświadczenia a wynagrodzeniem')
    plt.xlabel('Lata doświadczenia')
    plt.ylabel('Wynagrodzenie')
    plt.grid(True)
    plt.show()
else:
    print("Plik CSV nie zawiera kolumn 'Years of Experience' lub 'Salary'. Proszę sprawdź nazwy kolumn.")
    print("Dostępne kolumny to:", df.columns.tolist())
```

Powyższy kod najpierw wczytuje plik CSV do obiektu DataFrame za pomocą biblioteki `pandas`. Następnie, korzystając z `matplotlib.pyplot`, tworzy wykres punktowy, gdzie na osi X znajdują się 'Lata doświadczenia', a na osi Y 'Wynagrodzenie'. Przed wykreśleniem sprawdzana jest obecność tych kolumn.
